```python
from huggingface_hub import snapshot_download
from dotenv import load_dotenv
import os
load_dotenv()  

hf_token = os.getenv("hf_token")   

model_name = "facebook/opt-350m"

snapshot_download(
    repo_id=model_name,
    local_dir="../opt-350m",             # save one folder back
    local_dir_use_symlinks=False,
    resume_download=True,
    token=hf_token                       # not required for OPT, but safe
)

print("✅ OPT-350M downloaded successfully!")
```
