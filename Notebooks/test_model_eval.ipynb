```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
model_path = "../opt-350m"
tokenizer = AutoTokenizer.from_pretrained(model_path)
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device = "cpu"
model = AutoModelForCausalLM.from_pretrained(model_path).to(device)
```

```python
def response_checker(prompt):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    outputs = model.generate(**inputs, max_length=100)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Beautify response: split by newline and strip spaces
    formatted_lines = [line.strip() for line in response.split("\n") if line.strip()]
    beautified_response = "\n".join(f"- {line}" for line in formatted_lines)

    print(beautified_response)

    qa_dict = {}
    qa_dict["question"] = prompt
    qa_dict["response"] = beautified_response
    return qa_dict
```

```python
response_checker("What is the capital of France?")
```
